Escribe unha función que teña como argumento o valor máximo de n e calcule

$\displaystyle \prod^{\infty}_{n=1} \frac{4\,n^2}{4\,n^2-1}.$

Compara o resultado obtido co resultado exacto, $\displaystyle \frac{\pi}{2}.$

```python
def ejercicio3(n):
    resultado = 1;
    for i in [1..n]:
        op1 = 4 * (i*i)
        resultado *= (op1/(op1-1))
    return resultado
```

```python
abs(ejercicio3(500)-pi/2).n()
```

```python
def ex3(n):
    prod = 1.
    for i in [1..n]:
        part = 4*i^2
        prod *= part/(part-1)
    return prod
abs(ex3(500)-pi/2).n()
```

Escribe un código de multiplicación rusa en Sage.

```python
def mult_rusa(a,b):
    out=0;
    A=a;
    B=b;
    while(A>0):
        if(A % 2 == 1):
            out+=B;
        A //=2;
        B*=2
    return out
        
```

```python
mult_rusa(2,3)
```

```python
def mrusa(A,B):
    prod=0
    while A>0:
        if A%2==1:
            prod += B
        A//=2
        B*=2
    return prod
mrusa(27,82)
```

Dados tres vectores de dimensión 3, $\mathbf{u}$, $\mathbf{v}$ e $\mathbf{w}$, o seu producto mixto defínese por: 
$$[\mathbf{u},\mathbf{v},\mathbf{w}] = \mathbf{u}\cdot(\mathbf{v}\times \mathbf{w}).$$
Escribe unha función que calcule o producto mixto e escribe tamén como a usarías para comprobar nun caso particular a igualdade:
$$[\mathbf{u},\mathbf{w},\mathbf{v}] = -[\mathbf{u},\mathbf{v},\mathbf{w}].$$



```python
def mixto(u,v,w):
    return u.dot_product(v.cross_product(w))

u = vector(RDF, [1,2,3])
v = vector(RDF, [-4,5,6])
w = vector(RDF, [7,-8,9])
abs(mixto(u,w,v) + mixto(u,v,w)) < RDF.epsilon()*mixto(u,v,w)
```

```python
# Versión óptima sumando de un en un
def seriea4(tol):
    suma = 0.
    k = 1
    signo = 1.
    term = signo/(2*k-1)
    while abs(term) > tol:
        suma += term
        k += 1
        signo = -signo
        term =  signo/(2*k-1)
    return suma
seriea4(1e-3)
```

```python
# Versión óptima sumando de dous en dous
def seriea4b(tol):
    suma = 0.
    k = 1
    term = 1./(2*k-1)
    while term > tol:
        suma += term - 1./(2*k+1)
        k += 2
        term =  1./(2*k-1)
    return suma
seriea4b(1e-3)
```

A constante de Brun, $B_2$, é o valor ao que converxe a suma dos inversos dos números primos xemelgos:
$$B_2 = \left(\frac 13 + \frac 15\right) + \left(\frac 15 + \frac 17\right) + \left(\frac 1{11} + \frac 1{13}\right) + \left(\frac 1{17} + \frac 1{19}\right) + \left(\frac 1{29} + \frac 1{31}\right)+\ldots$$
Os primos xemelgos son aqueles primos separados por dúas unidades. Escribe unha función que constrúa a suma anterior para os primos menores que un dado, $N$. Escribe tamén o resultado que obtés para $N=1e6$.



```python
def brun(N):
    suma = 0.
    for q in primes(N):
        r = next_prime(q)
        if r-q==2:
            suma += 1./q + 1./r
    return suma
brun(1e6)
```

Unha tripla de enteiros positivos $(a,b,c)$ dise de Cardano se satisfai a condición:
$$\sqrt[3]{a + b \sqrt{c}} + \sqrt[3]{a - b \sqrt{c}} = 1$$
Por exemplo, (2,1,5) é unha tripla de Cardano. Fai unha función que devolva cantas triplas de Cardano hai para valores $a+b+c \leq 20$. A afectos de comprobación, son 2.

**Nota:** Como se indica nos apuntes da materia, para calcular $\sqrt[3]{x}$ cando $x$ pode ser negativo, debes usar`sqrt3 = lambda x: sign(x)*abs(x)^(1/3)`. 


```python
sqrt3 = lambda x: sign(x)*abs(x)^(1/3)
def f(n):
    cont = 0
    for a in (1..n):
        for b in (1..n-a):
            for c in (1..n-a-b):
                if abs(sqrt3(a+b*sqrt(c)) + sqrt3(a-b*sqrt(c)) - 1.) < RDF.epsilon():
                    cont += 1
    return cont
f(20)
```

O uso de `sqrt3` (ou `real_nth_root` en SageMath 9.7) é necesario porque o cálculo de potencias con exponente fraccionario e base negativa non é algo inmediato, como pode verse nos seguintes exemplos:

```python
-1.^(1/3)
```

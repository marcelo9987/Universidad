```python
sec = [5,2,2,8,-4,-1,2]
def funcionMediaMovil(lista, p):
    salida = []
    for i in range(len(lista)-(p-1)):
        suma =0
        for j in range(i,i+p):
            suma += lista[j]
        salida.append(suma/3)
    return salida
        
funcionMediaMovil(sec,3)
```

```python
def calculaTermino(x,k):
    a1 = pow((-1),k)
    a2 = pow(x,(2*k)+1)
    a = a1*a2
    b = 2*k + 1
    return a/b

def funcionSerie(x,tol):
    DEBUG = 1
    salida = 0
    k=0
    salida = calculaTermino(x,k)
    k+=1
    termino = salida
    while(abs(termino) > tol):
        termino = calculaTermino(x,k)
        salida += termino
        
        if(DEBUG):
            print("termino", termino)
            print("salida" , salida.n())
            sleep(1)
            
        k+=1
    return salida
        
```

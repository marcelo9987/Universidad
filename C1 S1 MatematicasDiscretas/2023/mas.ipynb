```python
def interpolate_lagrange(x,y,punto):
    n = len(x)
    valores = []
    for i in range(n):
        valor_temporal = 1
        for j in range(n):
            if(i!=j):
                valor_temporal *= (punto-x[j])/(x[i]-x[j])
        valores.append(valor_temporal*y[i])
    salida = 0
    for i in range(n):
        salida += valores[i]
    return salida
```

```python
def convBin(num):
    res=[]
    h=0
    while(h==0):
        if num%2==0:
            res.append(num%2)
        if(num%2!=0):
            res.append(num%2)
            
    return res
```

```python
def multiplicacionRusa(numero1,numero2):
    acc = 0
    res = 0
    count=0
    while(numero1>0):
        
        
        if(numero1%2 !=0):
            numero1 //=2
            acc+=numero2*pow(2,count)
            count+=1
            
        if(numero1%2==0):
            numero1 //=2
            count+=1
        
        
    res=acc
    return res
```

```python
def bnp50():
    res=[]
    pris=[]
    p=Primes()
    for w in range(50):
        pris.append(p[w])
        
    v=50
    while(v>1):
        v-=1
        if ((pris[v]-1) %3 ==0 and pris[v]<50):
                res.append(pris[v])
        
    
    return res
```

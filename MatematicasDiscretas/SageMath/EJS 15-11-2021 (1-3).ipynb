```python
def matrizDimension(dim):
    V=Matrix(dim,dim)
    for i in range(dim):
        for h in range(dim):
            if i == h:
                V[i,h]= 2
                if(i==0):
                    V[i,(h+1)]=-1
                elif(i==h==dim-1):
                    V[i,(h-1)]=-1
                else:
                    V[i,(h+1)]=-1
                    V[i,(h-1)]=-1
                    
    return V
```

```python
def ln2(errmax):
    suma=1.00
    valre=log(2).n()
    n=1
    while abs(suma-valre) > errmax:
        n+=1
        if(n%2==0):
            suma= suma - 1/n
            
        if(n%2!=0):
            suma = suma + 1/n
        
            
    return suma
```

```python
def multipPi(maxN):
    res=1.00
    n=1.00
    while(n != maxN):
        res*=(4*n*n)/(4*n*n-1)
        n+=1
    return res
        
    
```
